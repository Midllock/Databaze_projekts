### 1. Jaká data se nacházejí v atributu `description`?
Sloupec `description` je typu **JSON**. Tento formát umožňuje ukládat nestrukturovaná nebo částečně strukturovaná data přímo v tabulce.

**Struktura dat:**
* Data jsou uložena jako **objekty** (klíč: hodnota). 
* Obsahují technické specifikace, které se liší podle typu produktu (např. u notebooku najdeme `cpu` a `ram_gb`, u židle `ergonomic` a `color`).
* Téměř všechny záznamy obsahují společný klíč `"category"`, který slouží k logickému zařazení produktu.

### 2. Co provede dotaz s `JSON_VALUE`?
Dotaz extrahuje hodnotu z JSON dokumentu na základě zadané cesty.

```sql
SELECT
    name,
    JSON_VALUE(description, '$.category') AS category
FROM product;

````python
SELECT p.*, w.name AS warehouse_name
FROM product p
JOIN warehouse w ON p.warehouse_id = w.id
WHERE w.id = 1;
````

### 4. Produkty s kategorií 'cable' v popisu
Využíváme cestu `$.category` k filtraci přímo uvnitř JSON objektu.

````python
SELECT id, name, quantity, description
FROM product
WHERE JSON_EXTRACT(description, '$.category') = 'cable';

-- Alternativní (zkrácený) zápis v MySQL:
-- SELECT * FROM product WHERE description->>'$.category' = 'cable';
````

### 5. Produkty s počtem větším než 20 (seřazeno vzestupně)
Filtrujeme na základě sloupce `quantity` a řadíme pomocí `ORDER BY`.

````python
SELECT name, quantity
FROM product
WHERE quantity > 20
ORDER BY quantity ASC;
````

### 6. Produkty uskladněné v březnu 2024
K filtraci použijeme funkce `MONTH()` a `YEAR()` nad sloupcem `stored_from`.

````python
SELECT name, quantity, stored_from
FROM product
WHERE MONTH(stored_from) = 3 
  AND YEAR(stored_from) = 2024;
````
