# Databázový úkol: Sklad a produkty (JSON)
Tento notebook obsahuje řešení úkolů zaměřených na práci s MySQL databází a datovým typem JSON.

## 1. Analýza atributu description
**Otázka:** Jaká data se v něm nacházejí a jakou mají strukturu?

**Odpověď:**
Atribut `description` používá datový typ **JSON**. Obsahuje doplňkové parametry produktů ve formátu klíč-hodnota. 
* **Struktura:** Jedná se o nestrukturovaný objekt, kde se klíče liší podle typu produktu (např. u monitoru je to `size_inch`, u notebooku `cpu`).
* **Společné prvky:** Většina záznamů obsahuje klíč `category`, který definuje typ zboží.

## 2. Význam dotazu s JSON_VALUE
**Dotaz:**
```sql
SELECT
    name,
    JSON_VALUE(description, '$.category') AS category
FROM product;
```
**Vysvětlení:**
Tento dotaz vybere název produktu a z JSON objektu ve sloupci `description` vyextrahuje hodnotu uloženou pod klíčem `category`. Výsledek se zobrazí jako běžný textový sloupec s názvem `category`.

## 3. SQL: Produkty z vybraného skladu
Zobrazení všech produktů ze skladu 'Central Warehouse' (id = 1).

````python
-- SQL skript
SELECT p.*, w.name AS warehouse_name
FROM product p
JOIN warehouse w ON p.warehouse_id = w.id
WHERE w.id = 1;
````

## 4. SQL: Produkty v kategorii 'cable'
Vyhledání produktů, které mají v JSON popisu kategorii nastavenou na 'cable'.

````python
-- SQL skript
SELECT * 
FROM product 
WHERE description->>'$.category' = 'cable';
````

## 5. SQL: Produkty s počtem > 20
Seřazeno vzestupně podle počtu kusů.

````python
-- SQL skript
SELECT name, quantity 
FROM product 
WHERE quantity > 20 
ORDER BY quantity ASC;
````

## 6. SQL: Produkty uskladněné v březnu 2024

````python
-- SQL skript
SELECT * 
FROM product 
WHERE MONTH(stored_from) = 3 
  AND YEAR(stored_from) = 2024;
````
